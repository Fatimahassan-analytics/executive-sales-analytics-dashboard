```python
# Setup
from pathlib import Path
import sys, json
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
import numpy as np

from data import load_data
from model import add_derived_columns, kpis, sales_by_region, sales_by_category

ARTEFACTS_DIR = PROJECT_ROOT / 'artefacts'
df = add_derived_columns(load_data())
k = kpis(df)
print("KPI snapshot:")
for kk, vv in k.items():
    print(f"  {kk}: {vv}")
```

```python
# 1. Five business questions answered by the dashboard
questions = [
    ("What was total sales last week vs the previous week?",
     "KPI strip — WoW sales change metric, delta-coloured red/green."),
    ("Which regions are over- or under-performing?",
     "Bar chart of sales by region, coloured by profit."),
    ("Which product categories are driving changes?",
     "Stacked bar chart of sales by category and sub-category."),
    ("What is the impact of shipping mode / customer segment?",
     "Pie chart for segment + bar chart for ship mode."),
    ("How is YTD tracking against annual target?",
     "Progress bar comparing YTD sales to prior year's total."),
]
for q, a in questions:
    print(f"Q: {q}")
    print(f"A: {a}")
    print()
```

```python
# 2. Design rationale — why these KPIs and not 10?
lines = [
    "Why 5 KPIs, not 10?",
    "- A regional sales director's Monday-morning scan should be answerable in 30 seconds.",
    "- 5 KPIs fit on one screen row; 10 would force a second row that gets ignored.",
    "- Each KPI answers a distinct question; no overlap.",
]
print("\n".join(lines))
```

```python
# 3. Design rationale — why Plotly?
lines = [
    "Why Plotly over Altair or Bokeh?",
    "- Interactive hover, zoom, and filter without custom JavaScript.",
    "- Renders cleanly on Streamlit Cloud without extra config.",
    "- Plotly Express API matches pandas idioms closely, making the code readable.",
]
print("\n".join(lines))
```

```python
# 4. Design rationale — why stacked bar for category drill-down?
lines = [
    "Why a stacked bar for category drill-down?",
    "- A regional director can answer 'which sub-category is dragging down Office Supplies?'",
    "  in one glance, without clicking.",
    "- The colour legend doubles as a sub-category inventory list, saving screen real estate.",
]
print("\n".join(lines))
```

```python
# 5. Design rationale — why YTD vs prior-year total as the target?
lines = [
    "Why YTD vs prior-year total as the target?",
    "- Real sales targets are usually negotiated; without access to that data, prior-year",
    "  total is the most defensible benchmark.",
    "- It is also what most regional directors actually use as a sanity check.",
    "- If real targets become available, the metric can be parameterised via Streamlit sidebar.",
]
print("\n".join(lines))
```
