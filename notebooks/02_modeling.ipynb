```python
# Setup
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
import numpy as np

from data import load_data
from model import (add_derived_columns, kpis, sales_by_region, sales_by_category,
                   sales_by_segment, sales_over_time, sales_by_ship_mode)

ARTEFACTS_DIR = PROJECT_ROOT / 'artefacts'
ARTEFACTS_DIR.mkdir(exist_ok=True)
```

```python
# 1. Load data + derive columns
df = add_derived_columns(load_data())
print(f"Shape: {df.shape}")
```

```python
# 2. Test KPI computation
k = kpis(df)
import json
print(json.dumps(k, indent=2, default=str))
```

```python
# 3. Test regional aggregation
regional = sales_by_region(df)
print(regional.to_string(index=False))
```

```python
# 4. Test category aggregation
cat = sales_by_category(df)
print(cat.head(15).to_string(index=False))
```

```python
# 5. Test segment aggregation
seg = sales_by_segment(df)
print(seg.to_string(index=False))
```

```python
# 6. Test ship-mode aggregation
sm = sales_by_ship_mode(df)
print(sm.to_string(index=False))
```

```python
# 7. Test time-series aggregation at different frequencies
for freq in ['D', 'W', 'M', 'Q', 'Y']:
    s = sales_over_time(df, freq=freq)
    print(f"  {freq}: {len(s)} rows; last row: {s.iloc[-1].to_dict()}")
```

```python
# 8. Save KPI snapshot for the evaluation notebook
import json
out_path = ARTEFACTS_DIR / 'kpi_snapshot.json'
with open(out_path, 'w') as f:
    json.dump(k, f, indent=2, default=str)
print(f"Saved → {out_path}")
```
