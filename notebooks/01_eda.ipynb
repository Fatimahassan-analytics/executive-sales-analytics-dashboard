```python
# Setup
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from data import load_data, RAW_DIR
from model import add_derived_columns, kpis
from utils import (save_fig, plot_sales_over_time, plot_sales_by_region,
                   plot_sales_by_category, plot_profit_margin_by_segment)

pd.set_option('display.max_columns', None); pd.set_option('display.width', 120)
sns.set_style('whitegrid')
print(f"Raw data dir: {RAW_DIR}")
```

```python
# 1. Load data + add derived columns
df = add_derived_columns(load_data())
print(f"Shape: {df.shape}")
print(f"Date range: {df['OrderDate'].min()} to {df['OrderDate'].max()}")
df.head()
```

```python
# 2. Headline KPIs
k = kpis(df)
for kk, vv in k.items():
    if isinstance(vv, float):
        print(f"  {kk:25s}: {vv:,.2f}")
    else:
        print(f"  {kk:25s}: {vv}")
```

```python
# 3. Sales by region
plot_sales_by_region(df)
save_fig('01_sales_by_region.png')
plt.show()
print(df.groupby('Region')['Sales'].sum().sort_values(ascending=False).round(0))
```

```python
# 4. Sales by category (stacked by sub-category)
plot_sales_by_category(df)
save_fig('02_sales_by_category.png')
plt.show()
print(df.groupby(['Category','SubCategory'])['Sales'].sum().round(0))
```

```python
# 5. Sales and profit over time (monthly)
plot_sales_over_time(df, freq='M')
save_fig('03_sales_over_time.png')
plt.show()
```

```python
# 6. Profit margin by customer segment
plot_profit_margin_by_segment(df)
save_fig('04_profit_margin_by_segment.png')
plt.show()
```

```python
# 7. Save a cleaned copy for the modelling notebook
from data import save_processed
out = save_processed(df, name='superstore_clean.csv')
print(f"Saved → {out}")
```
